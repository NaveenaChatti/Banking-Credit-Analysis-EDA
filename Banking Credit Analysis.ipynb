```python
# dropping columns with more than 30% null values
null_columns=app_data.isnull().sum()
null_columns= list(null_columns[null_columns.values>=0.3].index)
app_data.drop(labels=null_columns,axis=1,inplace=True)
print(len(null_columns))
```

```python
app_data.isnull().sum()/len(app_data)*100
```

```python
null_columns1=prev_data.isnull().sum()
null_columns1= list(null_columns1[null_columns1.values>=0.3].index)
prev_data.drop(labels=null_columns1,axis=1,inplace=True)
print(len(null_columns1))
```

```python
app_data.isnull().sum()/len(app_data)*100
```

```python
# Removing rows having null values greater than or equal to 30%

null_row=app_data.isnull().sum(axis=0)
null_row=list(null_row[null_row.values>=0.3*len(app_data)].index)
app_data.drop(labels=null_row,axis=0,inplace=True)
print(len(null_row))
```

```python
null_row1=prev_data.isnull().sum(axis=0)
null_row1=list(null_row1[null_row1.values>=0.3*len(app_data)].index)
prev_data.drop(labels=null_row1,axis=0,inplace=True)
print(len(null_row1))
```

```python
# BINNING VARIABLES-AMT INCOME RANGE, CREDIT RANGE, AGE GROUP
bins = [0,25000,50000,75000,100000,125000,150000,175000,200000,225000,250000,275000,300000,325000,350000,375000,400000,425000,450000,475000,500000,10000000000]
slot = ['0-25000', '25000-50000','50000-75000','75000,100000','100000-125000', '125000-150000', '150000-175000','175000-200000',
       '200000-225000','225000-250000','250000-275000','275000-300000','300000-325000','325000-350000','350000-375000',
       '375000-400000','400000-425000','425000-450000','450000-475000','475000-500000','500000 and above']

app_data['AMT_INCOME_RANGE']=pd.cut(app_data['AMT_INCOME_TOTAL'],bins,labels=slot)
```

```python
bins = [0,150000,200000,250000,300000,350000,400000,450000,500000,550000,600000,650000,700000,750000,800000,850000,900000,1000000000]
slots = ['0-150000', '150000-200000','200000-250000', '250000-300000', '300000-350000', '350000-400000','400000-450000',
        '450000-500000','500000-550000','550000-600000','600000-650000','650000-700000','700000-750000','750000-800000',
        '800000-850000','850000-900000','900000 and above']

app_data['AMT_CREDIT_RANGE']=pd.cut(app_data['AMT_CREDIT'],bins=bins,labels=slots)
```

```python
#UNIVARIATE ANALYSIS USING DATA VISUALISATION

def uniplot(app_data,col,title,hue =None):
    
    sns.set_style('dark')
    sns.set_context('paper')
    plt.rcParams["axes.labelsize"] = 21
    plt.rcParams['axes.titlesize'] = 23
    plt.rcParams['axes.titlepad'] = 31
    
    
    temp = pd.Series(data = hue)
    fig, ax = plt.subplots()
    width = len(app_data[col].unique()) + 7 + 4*len(temp.unique())
    fig.set_size_inches(width , 8)
    plt.xticks(rotation=45)
    plt.yscale('log')
    plt.title(title)
    ax = sns.countplot(data = app_data, x= col, order=app_data[col].value_counts().index,hue = hue,palette='viridis') 
        
    plt.show()
```
